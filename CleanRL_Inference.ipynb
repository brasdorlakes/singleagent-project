```python


#This script was initially made with ChatGPT and then modified for our own purposes
import torch
import gymnasium as gym
import numpy as np
from Single_ag_Environment import SingleSatelliteEnv#Import our environment
from cleanrl.cleanrl.c51 import QNetwork#Import the QNetwork from the CleanRL implementation
env = SingleSatelliteEnv()
Testing_seed_Configuration=[5,6,7,8,9,10]#Test seeds
num_of_test_episodes=10#Number of test episodes

envs = gym.vector.SyncVectorEnv([
    lambda: SingleSatelliteEnv()#Need to setup a vectorized environment because that is what the CleanRL model expects
])
for i in range(0,5):
    for j in range(0,num_of_test_episodes):
        obs, _ = envs.reset(seed=Testing_seed_Configuration[i])#Reset using the current seed
        
        model = QNetwork(envs).to("cpu")#setup the model 
        
        # Load trained weights
        checkpoint = torch.load("cleanrl/cleanrl/runs/SingleSatelliteEnv__c51__1__1764003257/c51.cleanrl_model")#import the checkpoint
        #print(checkpoint)
        model.load_state_dict(checkpoint["model_weights"]) #CleanRL saves weights under "model"
        #print(dir(model))
        # Inference loop
        done = False
        while not done:
            obs_tensor = torch.tensor(obs, dtype=torch.float32)
            with torch.no_grad():
                action, q_values = model.get_action(obs_tensor)#Get the action value from the get action function
            
            obs, reward, terminated, truncated, info = envs.step(action)
            done = terminated or truncated
            
        env.close()
```
