```python
#This script was initially made with ChatGPT and then modified for our own purposes


import torch
import gymnasium as gym
import numpy as np
from Single_ag_Environment import SingleSatelliteEnv
from cleanrl.cleanrl.c51 import QNetwork
# Import the same policy class used in the CleanRL script
env = SingleSatelliteEnv()
# Create agent (must match training architecture!)
Testing_seed_Configuration=[5,6,7,8,9,10]
num_of_test_episodes=10
import gymnasium as gym

envs = gym.vector.SyncVectorEnv([
    lambda: SingleSatelliteEnv()
])
for i in range(0,5):
    for j in range(0,num_of_test_episodes):
        obs, _ = envs.reset(seed=Testing_seed_Configuration[i])
        
        model = QNetwork(envs).to("cpu")
        
        # Load trained weights
        checkpoint = torch.load("cleanrl/cleanrl/runs/SingleSatelliteEnv__c51__1__1764003257/c51.cleanrl_model")
        print(checkpoint)
        model.load_state_dict(checkpoint["model_weights"]) #CleanRL saves weights under "model"
        print(dir(model))
        # Inference loop
        done = False
        while not done:
            obs_tensor = torch.tensor(obs, dtype=torch.float32)
            with torch.no_grad():
                action, q_values = model.get_action(obs_tensor)
            
            obs, reward, terminated, truncated, info = envs.step(action)
            done = terminated or truncated
            
        env.close()
```
